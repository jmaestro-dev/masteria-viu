#### Resumen del artículo
##### Resumen: Capítulo 1 — Correlaciones en Datos Financieros

##### Idea Central
La idea de las correlaciones de datos financieros (retornos de los activos en un mismo período de tiempo) es la relación fundamental que gobierna la mayoría de las inversiones financieras. Si cada serie de datos financieros, como las tasas de retornos de las acciones, es probabilísticamente independiente de las demás acciones, entonces no existen factores de riesgo sistemáticos que afecten a todas las acciones; cada acción se mueve según su propio riesgo independiente. Esto último no refleja las interrelaciones reales entre el mercado bursátil y el sector corporativo. Las empresas cotizadas sí comparten riesgos comunes, como cuando pertenecen al mismo sector afectado por un cambio sectorial o cuando se ven afectadas por el coste de los fondos al variar los tipos de interés del mercado. Cuando la rentabilidad de las acciones se mueve conjuntamente o presenta correlaciones distintas de cero, pueden combinarse en una cartera de forma óptima para maximizar el beneficio del inversor.

##### 1. Revisión de herramientas Python
El capítulo arranca con un repaso práctico de **NumPy** (operaciones matriciales y arrays) y **Pandas** (manipulación de datos tabulares), que se usan a lo largo de todo el análisis.

##### 2. Diversificación de Portafolios (Markowitz)

Se presentan tres problemas de optimización bajo restricciones de no-venta en corto:
- **Frontera eficiente de Markowitz**: representa el límite de portafolios de acciones riesgosas que tienen el menor valor de varianza sobre los retornos dado el valor esperado de retorno para esa combinación de activos y que el retorno esperado sea mayor o igual al retorno del MVP (Minimum variance portfolio). El gráfico de retornos medios versus la volatilidad (o desvío estándar de los retornos, o riesgo) cuando no hay restricciones de ventas en corto tiene la forma de una hipérbola, y es usualmente llamado el portafolio de la frontera eficiente.
- **Máximo Sharpe Ratio**: maximizar el retorno esperado por unidad de riesgo.

El ejemplo práctico usa 7 empresas de e-commerce (Alibaba(BABA.US), Amazon(AMZN.US), eBay (EBAY.US), Rakuten (4755.jp), Wayfair (W.US), Zalando (zal.de), JD.com (JD.US)), con datos descargados del sitio https://stooq.com del período 2017–2021. Los retornos se calculan como logaritmos continuamente compuestos sobre precios ajustados por dividendos y splits. NOTA: he eliminado la empresa Suning incluída en el libro por no haber podido descargar los retornos de esta empresa japonesa.

##### 3. Portafolios Multi-sector
Se amplía el universo a 24 acciones en tres sectores (e-commerce, semiconductores, alimentos). Se demuestra que **la frontera eficiente combinada domina a cada sector por separado**: para igual retorno esperado, la volatilidad es menor — evidencia directa del beneficio de la diversificación.
Sin embargo, los pesos óptimos calculados en el conjunto de entrenamiento (2017–2020) **no se transfieren bien** al conjunto de test (2021), porque los retornos futuros difieren de los históricos. Esto ilustra el principal talón de Aquiles de la optimización clásica.


##### 4. Denoising de la Matriz de Correlación

El problema: con N activos y T observaciones finitas, la matriz de correlación muestral es una variable aleatoria, no la verdadera matriz poblacional.
El **Teorema de Marchenko–Pastur** establece que, si los datos fueran puro ruido, los autovalores de la matriz de correlación seguirían una distribución teórica acotada entre L y U. Los autovalores **por encima de U ≈ 1,3** (para N=24, T=889) representan señales genuinas del mercado; los que caen por debajo son ruido.
**Método de denoising** (constant residual eigenvalue): reemplazar los autovalores menores que U por su promedio (0,5719), preservando la varianza total, y reconstruir la matriz de covarianza. En este caso particular el beneficio fue marginal, probablemente porque el error en la estimación de los retornos esperados domina sobre el error en la covarianza.

##### 5. Predictor Prospectivo Más Preciso

La sección más importante conceptualmente: si se sustituyen los retornos históricos medios por **predicciones más precisas de los retornos futuros**, el portafolio out-of-sample se acerca considerablemente a la frontera eficiente in-sample. Esto demuestra que la calidad de la predicción de retornos es más determinante para el desempeño real que la diversificación en sí misma.

##### 6. Conclusiones

````python
import scipy
from scipy.optimize import Bounds
bounds = Bounds(0, 1)

from scipy.optimize import LinearConstraint
linear_constraint = LinearConstraint(np.ones((dfret.shape[1],), dtype=int), 1, 1)

covar = dfret.cov()
r = np.mean(dfret, axis=0) * 252

def ret(r, w):
    return r.dot(w)

def vol(w, covar):
    return np.sqrt(np.dot(w, np.dot(w, covar)) * 252)

def sharpe(ret, vol):
    return ret / vol

from scipy.optimize import minimize
weights = np.ones(dfret.shape[1])
x0 = weights / np.sum(weights)

portfstderr = lambda w: np.sqrt(np.dot(w, np.dot(w, covar)) * 252)
res1 = minimize(portfstderr, x0, method='trust-constr', constraints=linear_constraint, bounds=bounds)

w_min = res1.x
np.set_printoptions(suppress = True, precision=3)
print(w_min)
print('return: % .4f'% (ret(r,w_min)), 'risk: % .4f'% vol(w_min,covar))
````

````python
### Define 1/Sharpe_ratio as invSharpe
invSharpe = lambda w: np.sqrt(np.dot(w,np.dot(w,covar))*252)/r.dot(w)
res2 = minimize(invSharpe,x0,method='trust-constr',constraints = linear_constraint,bounds = bounds)
### Objective function is invSharpe -- inverse of Sharpe ratio
### These are the weights of the stocks in the portfolio with the highest Sharpe ratio
### -- call the weight vector w_Sharpe

w_Sharpe = res2.x
### constraint means unit vector .dot(w) = 1; minimize chooses wts w
### optimization full output.x gives the solution array of optimal weights in min inverse Sharpe
### ratio or max Sharpe ratio

print(w_Sharpe)
print('return: % .4f'% (ret(r,w_Sharpe)), 'risk: % .4f'% vol(w_Sharpe,covar))
### this is max Sharpe ratio portfolio

print( 1/ ( np.sqrt(np.dot(w_Sharpe,np.dot(w_Sharpe,covar))*252)/r.dot(w_Sharpe) ) )
### Above is optimized objective function -- the max Sharpe ratio.
### It can also be found using print(sharpe(ret(r,w_Sharpe),vol(w_Sharpe,covar)))
````

````python
decovar = covar.copy()

w = w_min ### w is now optimal portfolio weights, sum to 1
num_ports = 100
gap = (np.amax(r) - ret(r,w_min))/num_ports
### np.amax in numpy returns max in the array -- since weights sum to 1 and are
### bounded in (0,1), max portf ret is amax(r)
### The aboove range given by gap starts at ret given by Min Var Portf to Max of all mean returns

all_weights = np.zeros((num_ports, len(dfret.columns)))
### all_weights is 20 100 x 8 zero matrix
### Note: len(dfret.columns) is 8 -- there are 8 stocks here
### print(np.shape(all_weights)) gives (100,8) - same as print(all_weights.shape) that gives (100,8)

print("************************************************************")
###########################################################################################
### First note that in Python, a Tuple is a grouping of unnamed, ordered values that can be
### of different types; an array is a collection where elements' values can be changed, and
### are of a single type. ALL_weights is a 2D array of 100 rows each with 8 cols.
### all_weight[0] below is the first 1D sub-array that is 1st row
### Note: all_weights[0] is [0. 0. 0. 0. 0. 0. 0. 0.], i.e. first row of 100 x 8 all_weight
### Note: all_weights[1] is [0. 0. 0. 0. 0. 0. 0. 0.], i.e. second row of 100 x 8 all_weight
### ..........
### Note: all_weights[99 is [0. 0. 0. 0. 0. 0. 0. 0.], i.e. 100th row of 100 x 8 all_weight
### print(all_weights[0].shape) gives (8,) -- a 1-tuple with 8 elements
### print(all_weights.shape[0]) gives 100, i.e. dimension of the rows
### print(all_weights.shape[1]) gives 8, i.e. dim of the cols
###########################################################################################

ret_arr = np.zeros(num_ports) ### this is a 1-tuple of 100 zeros
### Note: print(ret_arr.shape) gives (100,) -- this is a 1D tuple
### print(ret_arr.shape[0]) gives 100, the number of elements in 1D tuple
### if we print(ret_arr.shape[1]) - this gives "tuple index out of range" as there is no other dim
### If we use instead ret_arr = np.zeros((num_ports,1)), then print(ret_arr.shape)
### gives (100,1), a dataframe with one column

vol_arr = np.zeros(num_ports)

for i in range(num_ports): ### this means looping from i=0 to 1,2,3,4,....,99 (100 loops in total)
    port_ret = ret(r,w) + i*gap
    double_constraint = LinearConstraint([np.ones(dfret.shape[1]),r],[1,port_ret],[1,port_ret])
### above means left term in first bracket np.ones(dfret.shape[1]) or A * w >< first term
### in 2nd,3rd brackets 1,1;
### right term in first bracket r * w >< second terms in 2nd, 3rd brackets port_ret
### i.e. constraints port wts sum to 1 and portf exp return == port_ret

### Create x0: initial guesses for weights.
    x0 = w_min
### Define a function for annualized portfolio volatility.
    portfvol = lambda w: np.sqrt(np.dot(w,np.dot(w,decovar))*252)
    res = minimize(portfvol,x0,method='trust-constr',constraints = double_constraint,bounds = bounds)
### Above double constraints mean unit vector .dot(w) = 1; r .dot(w) = port_ret; minimize chooses wts w
    all_weights[i,:]=res.x ### 24 x 1 optimal wts at row i, formed using decovar
    ret_arr[i]=port_ret
    vol_arr[i]=vol(res.x,decovar)

### Indented paras after "for i..." form the loop

sharpe_arr = ret_arr/vol_arr ### sharpe_arr is 100 x 1 array since it is ret_arr[100]/vol_arr[100] element by element

plt.figure(figsize=(20,10)) ### optimal frontier using training set and decovar
plt.scatter(vol_arr, ret_arr, c=sharpe_arr, cmap='viridis')
### in plt.scatter, c is a scalar or sequence of n numbers to be mapped to colors using cmap
### in plt, for sequential plots, 'viridis' gives colors across the 3D representation of vol_arr, ret_arr, sharpe_arr
### c= in front of third dimension sharpe_arr gives the colors in that dimension, otherwise dots will be all blue
plt.colorbar(label='Sharpe Ratio')
plt.xlabel('volatility')
plt.ylabel('Return')
plt.show()
````

##### Portafolios óptimo

Por último, se obtienen los pesos de las acciones de las 7 empresas que forman el portafolios óptimo. Estos pesos son la solución a un problema de minimización donde la función objetivo es el error standard del retorno anual del portafolio. Las medidas clave de retorno medio y desvío estandard de este retorno están expresados en valores anualizados y se calculan de la siguiente manera:
- para una acción determinada <sub>i</sub>, su retorno actualizado es 252 * $r^*_{it}$, donde $r^*_{it}$ es la tasa de retorno diaria y es considerada incondicionalmente estacionaria (con la misma media y varianza cada vez). Entonces, la media del retorno anual es 252 * E($r^*_{it}$) donde E(.) es el operador correspondiente a la Esperanza estadística y E($r^*_{it}$) es estimado usando 1/T $\sum_{t=1}^{T} r^*_{it}$ con los retornos diarios $r^*_{it}$ , t=1,2,...,T. La media del retorno anualizado es 252 1/T $\sum_{t=1}^{T} r^*_{it}$ donde 252 es el valor aproximado de días donde hay trading de acciones en los Estados Unidos.
- para una acción determinada <sub>i</sub>, donde i=1, 2, ..., N con un retorno diario r*<sub>t</sub> en el día t, cada elemento <sub>ij</sub> de la covarianza $r^*_{it}$ y $r^*_{ij}$ es estimado como \
1/(T-k) $\sum_{t=1}^{T} (r^*_{it}-\mu_i)(r^*_{jt}-\mu_j)$ \
donde k=1 para i=j, y k=2 para $i \neq j$.

La matriz covarianza puede ser denotada en la forma var($r^*_t$). La volatilidad del retorno del portafolios es entonces \
$\sqrt{252 x (w~T~var(r^*_t)~w)}$ \
o bien
```python
np.sqrt(np.dot(w,np.dot(w,covar))*252)
```
